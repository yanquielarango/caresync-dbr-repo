```python
dbutils.widgets.text("table_name", "")
dbutils.widgets.text("landing_timestamp", "")

table_name = dbutils.widgets.get("table_name")
landing_timestamp = dbutils.widgets.get("landing_timestamp")
```

```python
bronze_df = spark.read.table(f"adb_caresync.bronze.{table_name}").filter(f"landing_timestamp = '{landing_timestamp}'")
```

```python
from pyspark.sql import functions as F

updated_df = bronze_df.withColumn("insert_timestamp", F.current_timestamp()).withColumn("ingest_timestamp", F.current_timestamp())
```

```python
updated_df.write.mode("overwrite").saveAsTable(f"adb_caresync.silver.{table_name}")
```

```python
records_added = updated_df.count()
```

```python
dbutils.notebook.exit(str(records_added))
```
